```
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

X = np.genfromtxt("data.csv", delimiter=",", dtype=None)
y = np.random.choice(2, size=10)
```

```
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=2)
```

```
model = LogisticRegression()

# `model` parameter shadows the global `model` passed in below
def train_model(X_local_train, y_local_train, model):
    model.fit(X_local_train, y_local_train)

train_model(X, y, model)
```

```
def test_model(X_local_test, model):
    model.predict(X_local_test)

test_model(X_test, model)
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.impute import SimpleImputer

X = np.genfromtxt("data.csv", delimiter=",", dtype=None)
y = np.random.choice(2, size=10)
```

```python
# fill missing values using statistics of ALL rows - the leak
imputer = SimpleImputer(strategy="mean")
X_filled = imputer.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_filled, y, test_size=0.25, random_state=2)
```

```python
lr = LogisticRegression()
lr.fit(X_train, y_train)
print(lr.score(X_test, y_test))
```
